# Data Reading

```python
df = spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format", "parquet")\
    .option("cloudFiles.schemaLocation", f"s3://learn-databricks-project-e2e-1-bronze/checkpoint_{p_file_name}")\
    .load(f"s3://learn-databricks-project-e2e-1-source/{p_file_name}")
```

# Data Writing

```python
df.writeStream.format("delta")\
    .outputMode("append")\
    .option("checkpointLocation", f"s3://learn-databricks-project-e2e-1-bronze/checkpoint_{p_file_name}")\
    .option("path", f"s3://learn-databricks-project-e2e-1-bronze/{p_file_name}")\
    .trigger(once=True)\
    .start()   
```

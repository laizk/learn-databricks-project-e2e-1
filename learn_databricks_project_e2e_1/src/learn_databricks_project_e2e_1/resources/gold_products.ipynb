## DLT Pipeline

#### Streaming Table

```python
import dlt
from pyspark.sql.functions import *
from pyspark.types import *
```

```python
@dlt.table()
def DimProducts_stage():
    df = spark.readStream.table("learn_e2e_1.silver.products")
    return df
```

#### Streaming View

```python
@dlt.view
def DimProducts_view():
    df = spark.readStream.table("Live.DimProducts_stage")
    return df
```
